# a. - Analytical stellar density profile

$$\rho(r) = \rho_c \left [ 1 - \left (\frac{r}{R_\star} \right )^2 \right ]$$

where $\rho_c$ is the central density.

Also, the formulation for the mass equation (a form of the eqn on Slide 5 of the [Lecture](https://carlnotsagan.github.io/ast400a/lectures/week2/day_2.html))

$$
M(r) = \int^{r}_{0} 4 \pi r^2 \rho dr
$$
